```python
import sys
sys.path.append(r'C:\CareerShieldAI\backend\app')    # add backend folder to path

import pandas as pd
from resume_parser import (extract_skills, extract_education,
                           extract_experience, calculate_quality_score,
                           calculate_ats_score)        # import all functions

# --- Load cleaned resume dataset ---
df = pd.read_csv(r'C:\CareerShieldAI\datasets\resume_cleaned.csv')  # load data

# --- Test on first resume ---
sample_resume = df['Resume_str'][0]                  # get first resume text

print("Sample Resume Category:", df['Category'][0])
print("\nExtracted Skills:")
print(extract_skills(sample_resume))                 # show skills found

print("\nExtracted Education:")
print(extract_education(sample_resume))               # show education found

print("\nExtracted Experience:")
print(f"{extract_experience(sample_resume)} years")  # show years of experience

print("\nQuality Score:")
print(f"{calculate_quality_score(sample_resume)} / 100")  # show quality score
```

```python
# --- Test on 5 different resumes from different categories ---
sample_indices = [0, 120, 240, 360, 480]              # pick 5 different rows

for idx in sample_indices:                            # loop through each index
    resume_text = df['Resume_str'][idx]                # get resume text
    category = df['Category'][idx]                     # get actual category

    skills = extract_skills(resume_text)                # extract skills
    quality = calculate_quality_score(resume_text)      # calculate quality score
    experience = extract_experience(resume_text)        # extract experience

    print(f"Category: {category}")
    print(f"Skills found: {len(skills)} -> {skills[:5]}")  # show first 5 skills
    print(f"Experience: {experience} years")
    print(f"Quality Score: {quality}/100")
    print("-" * 50)                                     # separator line
```

```python
# --- Check why DESIGNER resume has programming skills ---
designer_text = df['Resume_str'][120]                 # get the designer resume
print("Designer resume preview:")
print(designer_text[:500])                             # show first 500 characters

print("\n" + "="*60)

# --- Test ATS matching: resume vs job description ---
sample_job_description = """
We are looking for a Python developer with experience in
SQL, Machine Learning, and AWS. Knowledge of Docker and
Git is a plus. Bachelor degree required.
"""

it_resume = df[df['Category']=='INFORMATION-TECHNOLOGY']['Resume_str'].iloc[0]  # get IT resume

ats_score = calculate_ats_score(it_resume, sample_job_description)  # calculate match

print(f"\nIT Resume vs Job Description ATS Match: {ats_score}%")
print(f"\nResume skills: {extract_skills(it_resume)}")
print(f"Job skills required: {extract_skills(sample_job_description)}")
```
