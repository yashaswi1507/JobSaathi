```python
import sys
sys.path.append(r'C:\CareerShieldAI\backend\app')    # add backend folder to path

import pandas as pd
from resume_parser import extract_skills              # import skill extractor
from career_recommender import recommend_careers, find_skill_gap  # import career functions

# --- Load a sample IT resume ---
resume_df = pd.read_csv(r'C:\CareerShieldAI\datasets\resume_cleaned.csv')  # load resumes
it_resume = resume_df[resume_df['Category']=='INFORMATION-TECHNOLOGY']['Resume_str'].iloc[5]  # pick one

# --- Extract skills from this resume ---
skills = extract_skills(it_resume)                    # get skills list
print("Resume Skills Found:")
print(skills)

# --- Get career recommendations ---
print("\nTop 5 Career Recommendations:")
recommendations = recommend_careers(skills, top_n=5)  # get top 5 matches

for i, rec in enumerate(recommendations, 1):           # loop through results
    print(f"\n{i}. {rec['job_title']}")
    print(f"   Match Score: {rec['match_score']}%")
    print(f"   Required Skills: {rec['required_skills'][:100]}...")
```

```python
# --- Check what this resume actually says ---
print("Full Resume Text:")
print(it_resume[:600])
```

```python
# --- Find skill gap for the top recommended role ---
top_role = recommendations[0]                          # get #1 recommendation

gap = find_skill_gap(skills, top_role['required_skills'])  # compare skills

print(f"Target Role: {top_role['job_title']}")
print(f"Match Score: {top_role['match_score']}%")

print(f"\nSkills You Already Have:")
print(gap['matching_skills'])

print(f"\nSkills You Need to Learn:")
print(gap['missing_skills'])
```

```python
# --- Test skill gap with a resume that has fewer skills ---

# simulate a junior developer resume with limited skills
junior_skills = ['python', 'html', 'css', 'git']     # basic skills only

# get career recommendations for this profile
junior_recs = recommend_careers(junior_skills, top_n=1)  # get top match
top_role = junior_recs[0]                              # pick top role

# find skill gap
gap = find_skill_gap(junior_skills, top_role['required_skills'])

print(f"Target Role: {top_role['job_title']}")
print(f"Match Score: {top_role['match_score']}%")
print(f"\nSkills You Already Have:")
print(gap['matching_skills'])
print(f"\nSkills You Need to Learn:")
print(gap['missing_skills'])
```

```python
# --- Test with a Finance resume to verify skill gap works across different domains ---
finance_resume = resume_df[resume_df['Category']=='FINANCE']['Resume_str'].iloc[0]

# extract skills
finance_skills = extract_skills(finance_resume)
print("Finance Resume Skills:")
print(finance_skills)

# get recommendations
finance_recs = recommend_careers(finance_skills, top_n=3)
print("\nTop 3 Career Recommendations:")
for i, rec in enumerate(finance_recs, 1):
    print(f"\n{i}. {rec['job_title']}")
    print(f"   Match Score: {rec['match_score']}%")

# skill gap for top role
top_role = finance_recs[0]
gap = find_skill_gap(finance_skills, top_role['required_skills'])
print(f"\nSkill Gap for: {top_role['job_title']}")
print(f"Already Have: {gap['matching_skills']}")
print(f"Need to Learn: {gap['missing_skills']}")
```
