```python
import sys
sys.path.append(r'C:\CareerShieldAI\backend\app')  # add backend folder to path

import pandas as pd
from preprocessing import add_fraud_features        # import our preprocessing functions

# --- Load cleaned fraud dataset ---
df = pd.read_csv(r'C:\CareerShieldAI\datasets\fraud_cleaned.csv')  # load data

# --- Apply all preprocessing features ---
df = add_fraud_features(df)                         # add new feature columns

# --- Check results ---
print("Columns after preprocessing:")
print(list(df.columns))                             # show all columns

print("\nSample of new features:")
print(df[['title', 'suspicious_salary',             # show new feature columns
          'missing_profile', 'scam_keywords',
          'fraudulent']].head(10))                  # first 10 rows

print("\nScam keyword stats:")
print(df['scam_keywords'].value_counts())           # how many jobs have scam keywords

print("\nSuspicious salary stats:")
print(df['suspicious_salary'].value_counts())       # how many jobs have suspicious salary
```

```python
# --- Check how fraud correlates with our new features ---

print("Fraud vs Suspicious Salary:")
# group by suspicious salary and show fraud rate
print(df.groupby('suspicious_salary')['fraudulent'].mean().round(3))

print("\nFraud vs Missing Profile:")
# group by missing profile and show fraud rate
print(df.groupby('missing_profile')['fraudulent'].mean().round(3))

print("\nFraud vs Scam Keywords:")
# group by scam keywords and show fraud rate
print(df.groupby('scam_keywords')['fraudulent'].mean().round(3))
```

```python
# --- Save fully preprocessed fraud dataset ---
df.to_csv(r'C:\CareerShieldAI\datasets\fraud_features.csv', index=False)  # save to file

print("Preprocessed fraud dataset saved.")
print(f"Shape: {df.shape}")                         # rows and columns
print(f"New feature columns added: suspicious_salary, missing_profile, scam_keywords")
```
